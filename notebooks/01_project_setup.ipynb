# 01. Project Setup

This notebook initializes the project structure, checks the Python environment, and defines the main folders for the medical imaging pipeline.

```python
from pathlib import Path
import os

project_root = Path.cwd().resolve().parent
notebooks_dir = project_root / 'notebooks'
data_dir = project_root / 'data'
models_dir = project_root / 'models'
reports_dir = project_root / 'reports'

for folder in [notebooks_dir, data_dir, models_dir, reports_dir]:
    folder.mkdir(exist_ok=True, parents=True)

print(f'Project root: {project_root}')
print(f'Notebooks: {notebooks_dir}')
print(f'Data: {data_dir}')
print(f'Models: {models_dir}')
print(f'Reports: {reports_dir}')
```
