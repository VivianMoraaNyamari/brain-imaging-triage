# 05. Feature Engineering

Extract radiomic and handcrafted features from CT and MRI for conventional machine learning baselines.

```python
import numpy as np
import pandas as pd

features = [
    'mean_intensity',
    'std_intensity',
    'texture_contrast',
    'texture_entropy',
    'area',
    'eccentricity',
    'skewness',
    'kurtosis'
]

feature_df = pd.DataFrame(columns=features + ['label'])
print(feature_df.head())
```

```python
def compute_example_features(image_array):
    mean_intensity = np.mean(image_array)
    std_intensity = np.std(image_array)
    return {
        'mean_intensity': mean_intensity,
        'std_intensity': std_intensity,
        'texture_contrast': float(np.var(image_array)),
        'texture_entropy': float(np.mean(np.abs(np.diff(image_array.reshape(-1)))))
    }

print('Feature extraction function prepared.')
```
