# 02. Data Collection

This notebook defines the dataset structure for CT and MRI brain images and outlines how to organize labels for stroke, hemorrhage, and mass effect.

```python
from pathlib import Path

project_root = Path.cwd().resolve().parent
data_dir = project_root / 'data'

dataset_dirs = {
    'train': data_dir / 'train',
    'val': data_dir / 'val',
    'test': data_dir / 'test',
    'labels': data_dir / 'labels'
}

for path in dataset_dirs.values():
    path.mkdir(parents=True, exist_ok=True)

for label in ['stroke', 'hemorrhage', 'mass_effect', 'normal']:
    (dataset_dirs['train'] / label).mkdir(exist_ok=True, parents=True)
    (dataset_dirs['val'] / label).mkdir(exist_ok=True, parents=True)
    (dataset_dirs['test'] / label).mkdir(exist_ok=True, parents=True)

print(dataset_dirs)
```
