# 04. Preprocessing

Standardize CT and MRI images before training: resize, normalize, convert to grayscale if needed, and split data.

```python
from pathlib import Path
import numpy as np
from PIL import Image

project_root = Path.cwd().resolve().parent
data_dir = project_root / 'data'
processed_dir = project_root / 'processed'
processed_dir.mkdir(exist_ok=True, parents=True)

def standardize_image(image_path, size=(224, 224)):
    img = Image.open(image_path).convert('L')
    img = img.resize(size)
    arr = np.asarray(img, dtype=np.float32)
    arr = (arr - arr.min()) / (arr.max() - arr.min() + 1e-8)
    return arr

print('Preprocessing pipeline ready.')
print(f'Processed output directory: {processed_dir}')
```
