# 09. Explainability and Triage Prototype

Use SHAP and Grad-CAM to interpret model predictions and implement a basic triage decision layer.

```python
import numpy as np

label_map = {
    0: 'stroke',
    1: 'hemorrhage',
    2: 'mass_effect'
}

def triage_decision(prediction, confidence):
    if prediction == 'hemorrhage' and confidence > 0.7:
        return 'Immediate prioritization'
    elif prediction == 'mass_effect' and confidence > 0.65:
        return 'Urgent review'
    elif prediction == 'stroke' and confidence > 0.7:
        return 'High-priority review'
    return 'Routine review with specialist confirmation'

print('Triage function ready.')
```
