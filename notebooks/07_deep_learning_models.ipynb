# 07. Deep Learning Models

Train transfer-learning networks for CT and MRI classification using ResNet, DenseNet, and EfficientNet.

```python
import tensorflow as tf
from tensorflow.keras.applications import ResNet50, DenseNet121, EfficientNetB0

models = {
    'resnet50': ResNet50(weights='imagenet', include_top=False, input_shape=(224,224,3)),
    'densenet121': DenseNet121(weights='imagenet', include_top=False, input_shape=(224,224,3)),
    'efficientnetb0': EfficientNetB0(weights='imagenet', include_top=False, input_shape=(224,224,3))
}

print('Transfer learning models initialized.')
for key in models:
    print('-', key)
```
