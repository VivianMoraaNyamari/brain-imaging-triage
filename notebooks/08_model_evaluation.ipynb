# 08. Model Evaluation

Evaluate and compare model performance with accuracy, precision, recall, F1-score, ROC-AUC, and confusion matrices.

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
import numpy as np

def summarize_results(y_true, y_pred, y_proba=None, labels=None):
    print(classification_report(y_true, y_pred, target_names=labels))
    print('Confusion Matrix:')
    print(confusion_matrix(y_true, y_pred))
    if y_proba is not None:
        try:
            print('ROC-AUC:', roc_auc_score(y_true, y_proba, multi_class='ovr', labels=labels))
        except Exception as e:
            print('ROC-AUC not available for this case:', e)

print('Evaluation metrics function ready.')
```
